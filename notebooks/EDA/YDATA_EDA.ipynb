```python
import pandas as pd
from ydata_profiling import ProfileReport
import os
```

```python
DATA_PATH = os.path.join('../..data/raw/','raw_test_call_data.csv')
df = pd.read_csv(DATA_PATH)
df.shape
```

```python
# create and display a ydata profile for the dataframe
profile = ProfileReport(df, title="Call Data Profile", explorative=True)
profile.to_notebook_iframe()

# also save to an HTML file
out_dir = os.path.join('..', '..', 'reports')
os.makedirs(out_dir, exist_ok=True)
profile.to_file(os.path.join(out_dir, 'call_data_profile.html'))
```
